```python
import random
import time

while True:
    limit1 = int(input(" (Enter the number of digits for first number: "))
    if limit1 > 0:
        break
    print("Please enter numbers greater than 0")

while True:
    limit2 = int(input("Enter the number of digits for second number: "))
    if limit2 > 0:
        break
    print("Please enter numbers greater than 0")

operator = input("Enter Operator (+, -, *, /): ")

# Score Variables
right = 0
wrong = 0

right_time_total = 0
wrong_time_total = 0

total_questions = 0

# Fastest & Slowest Time Variables
fastest_time = None
slowest_time = None

print("\n--- Speed Math Practice Started ---")
print("Type 'result' anytime to stop and see final score.\n")

while True:

    # Random numbers
    min1 = 10 ** (limit1 - 1)
    max1 = (10 ** limit1) - 1

    min2 = 10 ** (limit2 - 1)
    max2 = (10 ** limit2) - 1

    num1 = random.randint(min1, max1)
    num2 = random.randint(min2, max2)

    # Operator Logic
    if operator == "+":
        correct_answer = num1 + num2

    elif operator == "-":
        correct_answer = num1 - num2

    elif operator == "*":
        correct_answer = num1 * num2

    elif operator == "/":
        correct_answer = round(num1 / num2, 2)

    else:
        print("Invalid Operator!")
        break

    # Start Timer
    start_time = time.time()

    # User Input
    user_input = input(f"{num1} {operator} {num2} = ")

    # Stop Program
    if user_input.lower() == "result":
        break

    # End Timer
    end_time = time.time()

    # Time Taken
    time_taken = round(end_time - start_time, 2)

    total_questions += 1

    # Fastest Time Update
    if fastest_time is None or time_taken < fastest_time:
        fastest_time = time_taken

    # Slowest Time Update
    if slowest_time is None or time_taken > slowest_time:
        slowest_time = time_taken

    # Answer Checking
    try:
        user_answer = float(user_input)

        if user_answer == correct_answer:
            print(f"✅ Well Done | Time Taken: {time_taken} sec\n")

            right += 1
            right_time_total += time_taken

        else:
            print("❌ Wrong Answer")
            print(f"Correct Answer = {correct_answer}")
            print(f"Time Taken: {time_taken} sec\n")

            wrong += 1
            wrong_time_total += time_taken

    except ValueError:
        print("⚠ Invalid Input! Please enter a number.\n")

        wrong += 1
        wrong_time_total += time_taken

# Final Scoreboard
print("\n========== FINAL RESULT ==========")

print(f"Total Questions : {total_questions}")
print(f"Right Answers   : {right}")
print(f"Wrong Answers   : {wrong}")

# Average Times
if right > 0:
    right_avg = round(right_time_total / right, 2)
else:
    right_avg = 0

if wrong > 0:
    wrong_avg = round(wrong_time_total / wrong, 2)
else:
    wrong_avg = 0

print(f"\n✅ Right Answer Average Time : {right_avg} sec")
print(f"❌ Wrong Answer Average Time : {wrong_avg} sec")

# Fastest & Slowest Result
if fastest_time is not None:
    print(f"\n⚡ Fastest Answer Time : {fastest_time} sec")

if slowest_time is not None:
    print(f"🐢 Slowest Answer Time : {slowest_time} sec")

print("==================================")
```
